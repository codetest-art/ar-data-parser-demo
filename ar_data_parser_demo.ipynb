```python
import json
def parse_ar_payload(payload):
  return {k: v for k, v in json.loads(payload).items() if 'path' in k}
```

```python
payload = '{"user_path": "/start", "image": "scene.jpg", "nav_path": "/next"}'
print(parse_ar_payload(payload))
```
